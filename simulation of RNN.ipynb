```python
import numpy as np
np.random.seed(858)
```

```python
def sigmoid(array1):
    array = np.array(array1)
    sigmoid_array = np.zeros(shape=(array.shape[0],array.shape[1]))
    for i in range(array.shape[0]):
        for j in range(array.shape[1]):
            sigmoid_array[i][j] = 1/(1 + np.exp(-array[i][j]))  
            
    return sigmoid_array
```

# data

```python
x = np.array([[-1.4293882e-04,  5.0827850e+01,  5.3502022e+01,  5.3334230e-01,
         5.4525346e-01, -7.5891501e-01],
       [-1.2225091e-02,  4.6592384e+01,  5.0203579e+01,  3.9558709e-01,
         4.1271874e-01, -8.5964209e-01]])
```

```python
x.shape
```

# parameter layer 1

```python
hidden_layer1 = 4
```

```python
wx = np.random.uniform(size=(6,hidden_layer1))
wh = np.random.uniform(size=(hidden_layer1,hidden_layer1))
b = np.random.uniform(size=(1,hidden_layer1))
Ht_min1 = np.zeros((1,hidden_layer1))
```

```python
Ht1_l1 = x[0]@wx +  Ht_min1@wh + b
Ht1_l1 = np.tanh(Ht1_l1)

Ht2_l1 = x[1]@wx + Ht1_l1@wh + b
Ht2_l1 = np.tanh(Ht2_l1)
```

```python
print(Ht1_l1.shape,Ht2_l1.shape)
```

# parameter layer 2

```python
hidden_layer2 = 2
```

```python
wx = np.random.uniform(size=(hidden_layer1,hidden_layer2))
wh = np.random.uniform(size=(hidden_layer2,hidden_layer2))
b = np.random.uniform(size=(1,hidden_layer2))
Ht_min1 = np.zeros((1,hidden_layer2))
```

```python
Ht1_l2 = Ht1_l1@wx +  Ht_min1@wh + b
Ht1_l2 = np.tanh(Ht1_l2)

Ht2_l2 = Ht2_l1@wx + Ht1_l2@wh + b
Ht2_l2 = np.tanh(Ht2_l2)
```

```python
print(Ht1_l2.shape,Ht2_l2.shape)
```

# Output

```python
whq = np.random.uniform(size=(hidden_layer2,1))
bq = np.random.uniform(size=(1,1))
```

```python
Ot1 = Ht1_l2@whq + bq
Ot2 = Ht2_l2@whq +bq
```
