```python
import numpy as np
np.random.seed(858)
```

```python
def sigmoid(array1):
    array = np.array(array1)
    sigmoid_array = np.zeros(shape=(array.shape[0],array.shape[1]))
    for i in range(array.shape[0]):
        for j in range(array.shape[1]):
            sigmoid_array[i][j] = 1/(1 + np.exp(-array[i][j]))  
            
    return sigmoid_array
```

# data

```python
x = np.array([[-1.4293882e-04,  5.0827850e+01,  5.3502022e+01,  5.3334230e-01,
         5.4525346e-01, -7.5891501e-01],
       [-1.2225091e-02,  4.6592384e+01,  5.0203579e+01,  3.9558709e-01,
         4.1271874e-01, -8.5964209e-01]])
```

```python
x.shape
```

# parameter layer 1

```python
hidden_layer1 = 4
```

## Left to Right

```python
#lr = left to right
# rl = right to left
```

```python
wx_lr = np.random.uniform(size=(6,hidden_layer1))
wh_lr = np.random.uniform(size=(hidden_layer1,hidden_layer1))
b_lr = np.random.uniform(size=(1,hidden_layer1))
Ht_min1 = np.zeros((1,hidden_layer1))
```

```python
Ht1_l1_lr = x[0]@wx_lr +  Ht_min1@wh_lr + b_lr
Ht1_l1_lr = np.tanh(Ht1_l1_lr)

Ht2_l1_lr = x[1]@wx_lr + Ht1_l1_lr@wh_lr + b_lr
Ht2_l1_lr = np.tanh(Ht2_l1_lr)
```

## right to left

```python
wx_rl = np.random.uniform(size=(6,hidden_layer1))
wh_rl = np.random.uniform(size=(hidden_layer1,hidden_layer1))
b_rl = np.random.uniform(size=(1,hidden_layer1))
Ht_min1 = np.zeros((1,hidden_layer1))
```

```python
Ht1_l1_rl = x[1]@wx_rl +  Ht_min1@wh_rl + b_rl
Ht1_l1_rl = np.tanh(Ht1_l1_rl)

Ht2_l1_rl = x[0]@wx_rl + Ht1_l1_rl@wh_rl + b_rl
Ht2_l1_rl = np.tanh(Ht2_l1_rl)
```

```python
print(Ht1_l1_rl.shape,Ht2_l1_rl.shape)
```

## concat the result

```python
Ht1_l1 = np.concatenate((Ht1_l1_lr,Ht1_l1_rl),axis=1)

Ht2_l1 = np.concatenate((Ht2_l1_lr,Ht2_l1_rl),axis=1)
print(Ht1_l1.shape,Ht2_l1.shape)
```

# parameter layer 2

```python
hidden_layer2 = 2
```

## left to right (1 to N)

```python
wx2_lr = np.random.uniform(size=(hidden_layer1*2,hidden_layer2))
wh2_lr = np.random.uniform(size=(hidden_layer2,hidden_layer2))
b2_lr = np.random.uniform(size=(1,hidden_layer2))
Ht_min1 = np.zeros((1,hidden_layer2))
```

```python
Ht1_l2_lr = Ht1_l1@wx2_lr +  Ht_min1@wh2_lr + b2_lr
Ht1_l2_lr = np.tanh(Ht1_l2_lr)

Ht2_l2_lr = Ht2_l1@wx2_lr + Ht1_l2_lr@wh2_lr + b2_lr
Ht2_l2_lr = np.tanh(Ht2_l2_lr)
```

## right to left (N to 1)

```python
wx2_rl = np.random.uniform(size=(hidden_layer1*2,hidden_layer2))
wh2_rl = np.random.uniform(size=(hidden_layer2,hidden_layer2))
b2_rl = np.random.uniform(size=(1,hidden_layer2))
Ht_min1 = np.zeros((1,hidden_layer2))
```

```python
Ht1_l2_rl = Ht2_l1@wx2_rl +  Ht_min1@wh2_rl + b2_rl
Ht1_l2_rl = np.tanh(Ht1_l2_rl)

Ht2_l2_rl = Ht1_l1@wx2_rl + Ht1_l2_rl@wh2_rl + b2_rl
Ht2_l2_rl = np.tanh(Ht2_l2_rl)
```

```python
Ht1_l2 = np.concatenate((Ht1_l2_lr,Ht1_l2_rl),axis=1)
Ht2_l2 = np.concatenate((Ht2_l2_lr,Ht2_l2_rl),axis=1)
```

```python
print(Ht1_l2.shape,Ht2_l2.shape)
```

# Output

```python
whq = np.random.uniform(size=(hidden_layer2*2,1))
bq = np.random.uniform(size=(1,1))
```

```python
Ot1 = Ht1_l2@whq + bq
Ot2 = Ht2_l2@whq +bq
```
